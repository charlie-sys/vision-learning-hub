## 2. 手算与数值检查
先预测输出，再运行；改变参数比较解析梯度与有限差分。


```python
import numpy as np
x, w, b, target = 2., 3., 1., 5.
def loss(weight):
    return (weight*x + b - target)**2 / 2
h = 1e-5
analytic = (w*x + b - target)*x
numeric = (loss(w+h)-loss(w-h))/(2*h)
print('loss:', loss(w))
print('analytic gradient:', analytic)
print('numeric gradient:', numeric)
print('close:', np.isclose(analytic, numeric))
```

## 3. 高斯分布与KL
样本均值/方差只近似理论值，增加样本量观察误差。


```python
import numpy as np
rng = np.random.default_rng(7)
mu, sigma = 1., 2.
kl = .5*(mu**2 + sigma**2 - 1 - np.log(sigma**2))
z = mu + sigma*rng.standard_normal(20000)
print('KL(q || N(0,1)):', kl)
print('sample mean / variance:', z.mean(), z.var())
print('expected mean / variance:', mu, sigma**2)
```

## 4. SciPy全局指派


```python
import numpy as np
from scipy.optimize import linear_sum_assignment
C = np.array([[4, 1, 3], [2, 0, 5], [3, 2, 2]])
rows, cols = linear_sum_assignment(C)
print('row -> col:', list(zip(rows.tolist(), cols.tolist())))
print('total cost:', C[rows, cols].sum())
print('试着修改一个代价，再比较全局最优解。')
```

## 5. 符号求导


```python
import sympy as sp
w, x, b, target = sp.symbols('w x b target')
loss = (w*x+b-target)**2 / 2
grad = sp.diff(loss, w)
print('loss:', loss)
print('dL/dw:', sp.factor(grad))
print('at x=2,w=3,b=1,target=5:', grad.subs({x:2,w:3,b:1,target:5}))
```

## 6. Brownian方差与Euler误差
图像应接近理论线，但不会完全重合。


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(7)
h, steps = .01, 100
paths = np.cumsum(rng.standard_normal((5000, steps))*np.sqrt(h), axis=1)
t = np.arange(1, steps+1)*h
print('Brownian terminal variance:', paths[:, -1].var())
print('Euler x(1):', (1-h)**steps, 'exact:', np.exp(-1))
plt.figure(figsize=(6, 3))
plt.plot(t, paths.var(axis=0), label='empirical variance')
plt.plot(t, t, '--', label='theory: Var(W_t)=t')
plt.xlabel('time')
plt.legend()
plt.tight_layout()
plt.show()
```
